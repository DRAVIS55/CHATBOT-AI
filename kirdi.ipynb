# **Mode Evaluation test**

## **Linear regression and random classification model test**

## **Data loading**

```python
import pandas as pd
ds=pd.read_csv("https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv")
```

## **Data separation**

```python
y=ds["logS"]
x=ds.drop("logS", axis=1)
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y, train_size=0.2, random_state=100)
```

# **Model Training**

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xtrain,ytrain)
```

## **Model Prediction**

```python
lr_pred_train=lr.predict(xtrain)
lr_pred_test=lr.predict(xtest)
```

## **Models Performance**

```python
from sklearn.metrics import mean_squared_error,r2_score
lr_mean_sqr_train=mean_squared_error(ytrain,lr_pred_train)
lr_mean_sqr_test=mean_squared_error(ytest,lr_pred_test)
lr_r2_train=r2_score(ytrain,lr_pred_train)
lr_r2_test=r2_score(ytest,lr_pred_test)
```

# **Data Simulation**

```python
lr_df=pd.DataFrame({"Linear Regression",lr_mean_sqr_test,lr_mean_sqr_train,lr_r2_test,lr_r2_train}).transpose()
lr_df.columns=["Mean Squared Error(Test)","Mean Squared Error(Train)","R2 Score(Test)","R2 Score(Train)","Model"]
```

## **Random Forest model**

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(max_depth=2, random_state=100)
rf.fit(xtrain,ytrain)
```

## **Random Forest Prediction**

```python
rf_pred_train=rf.predict(xtrain)
rf_pred_test=rf.predict(xtest)
```

### **Random Forest Valuation**

```python
rf_mean_sqr_train=mean_squared_error(ytrain,rf_pred_train)
rf_mean_sqr_test=mean_squared_error(ytest,rf_pred_test)
rf_r2_train=r2_score(ytrain,rf_pred_train)
rf_r2_test=r2_score(ytest,rf_pred_test)
rf_df=pd.DataFrame({"Random Forest",rf_mean_sqr_test,rf_mean_sqr_train,rf_r2_test,rf_r2_train}).transpose()
rf_df.columns=["Mean Squared Error(Test)","Mean Squared Error(Train)","R2 Score(Test)","R2 Score(Train)","Model"]
```

# **Comparing the two models**

```python
results=pd.concat([lr_df,rf_df],axis=0).reset_index()
results.drop("index", axis=1, inplace=True)
results
```
